```python
items = dbutils.fs.ls('/databricks-datasets/')
# Count folders and files in a single pass
folder_count = file_count = 0
for item in items:
    if item.name.endswith("/"):
        folder_count += 1
    else:
        file_count += 1
# Display the results
print(f"Total Folders: {folder_count}")
print(f"Total Files: {file_count}")
```
