## Creating views for the data

```python
%sql
create or replace view gizmobox.bronze.v_customers as
select
  *,
  _metadata.file_path as file_path
from
  json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers`
```

```python
%sql
select * from gizmobox.bronze.v_customers;
```

### Temprory views
They gets destroyed when a spark session ends.

```python
%sql
create or replace temporary view tv_customers as
select
  *,
  _metadata.file_path as file_path
from
  json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers`
```

```python
%sql
select * from tv_customers;
```

### Global temporary view
They gets dropped when the cluster gets termineted.
It does not work in serverless compute, as serverless compute server never gets terminted.

```python
%sql
create or replace global temporary view gtv_customers as
select
  *,
  _metadata.file_path as file_path
from
  json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers`
```
