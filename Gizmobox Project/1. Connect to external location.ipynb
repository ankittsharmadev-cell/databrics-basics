##1. Access the gizmobox container

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databrics_ext_dl_gizmobox
  URL 'abfss://gizmobox@databrics1extdl.dfs.core.windows.net/'
  WITH (STORAGE CREDENTIAL databrics_course_ext_sc)
  COMMENT 'connect to external storage'
```

##2. Create the catalog for gizmobox

```python
%sql
show catalogs;
```

```python
%sql
CREATE CATALOG IF NOT EXISTS gizmobox
      MANAGED LOCATION 'abfss://gizmobox@databrics1extdl.dfs.core.windows.net/'
      COMMENT 'catalog for gizmobox'
```

##3. Create schema for gimobox

```python
%sql
use catalog gizmobox;
    
CREATE SCHEMA IF NOT EXISTS landing
      MANAGED LOCATION 'abfss://gizmobox@databrics1extdl.dfs.core.windows.net/landing';

CREATE SCHEMA IF NOT EXISTS bronze
      MANAGED LOCATION 'abfss://gizmobox@databrics1extdl.dfs.core.windows.net/bronze';

CREATE SCHEMA IF NOT EXISTS silver
      MANAGED LOCATION 'abfss://gizmobox@databrics1extdl.dfs.core.windows.net/silver';

CREATE SCHEMA IF NOT EXISTS gold
      MANAGED LOCATION 'abfss://gizmobox@databrics1extdl.dfs.core.windows.net/gold'

```

```python
%sql
SHOW SCHEMAS
```

##4. Create volume
This is a really importent step as it will connect the folders to the databrics catalog and then we don't need to give complete path of the folder

```python
%sql
use catalog gizmobox;
use schema landing;


CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://gizmobox@databrics1extdl.dfs.core.windows.net/landing/operational_data/'
    COMMENT 'volume for gizmobox' 
```
