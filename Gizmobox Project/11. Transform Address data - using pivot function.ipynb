```python
%sql
SELECT customer_id,
       address_type,
       address_line_1,
       city,
       state,
       postcode
FROM gizmobox.bronze.v_addresses
```

```python
%sql
select
  *
from
  (
    SELECT
      customer_id,
      address_type,
      address_line_1,
      city,
      state,
      postcode
    FROM
      gizmobox.bronze.v_addresses
  )
    pivot (
      max(address_line_1) as address_line_1,
      max(city) as city,
      max(state) as state,
      max(postcode) as postcode for address_type in ('shipping', 'billing')
    )
```

```python
%sql
create table if not exists gizmobox.silver.addresses as
select
  *
from
  (
    SELECT
      customer_id,
      address_type,
      address_line_1,
      city,
      state,
      postcode
    FROM
      gizmobox.bronze.v_addresses
  )
    pivot (
      max(address_line_1) as address_line_1,
      max(city) as city,
      max(state) as state,
      max(postcode) as postcode for address_type in ('shipping', 'billing')
    )
```

```python
%sql
select * from gizmobox.silver.addresses
```
