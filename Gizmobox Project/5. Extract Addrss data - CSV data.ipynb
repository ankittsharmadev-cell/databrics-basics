## Extract data from address table which do not have comma seperated value, it has tab seperated value.

```python
%sql
select * from csv.`/Volumes/gizmobox/landing/operational_data/addresses/`
```

### To overcome this limitation we will use read_files function.

```python
%sql
select
  *
from
  read_files(
    '/Volumes/gizmobox/landing/operational_data/addresses',
    format => 'csv',
    delimiter => '\t',
    header => 'true'
  );
```
