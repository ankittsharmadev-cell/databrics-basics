## Query Orders Data as JSON Strings

1. Extract Top Level Column Values
2. Extract Array elements
3. Extract Nested Column Values
4. CAST Column Values to a Specific Data Type

```python
%sql
select * from gizmobox.bronze.v_orders
```

```python
%sql
select value: order_id, value from gizmobox.bronze.v_orders
```

```python
%sql
select value: items, value from gizmobox.bronze.v_orders
```

```python
%sql
select
  value:items[0] as item_1,
  value:items[1] as item_2,
  value
from
  gizmobox.bronze.v_orders
```

```python
%sql
select
  value:items[0].item_id :: int as item_1_id,
  value:items[0] as item_1,
  value:items[1] as item_2,
  value
from
  gizmobox.bronze.v_orders
```
