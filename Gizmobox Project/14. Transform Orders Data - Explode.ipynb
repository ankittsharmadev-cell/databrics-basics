## Transform Orders Data - Explode Arrays

1. Access elements from the JSON object
2. Deduplicate Array Elements
3. Explode Arrays
4. Write the Transformed Data to Silver Schema

```python
%sql
select * from gizmobox.silver.orders_json
```

```python
%sql
create or replace view gizmobox.silver.tv_item_exploded as
SELECT json_value.order_id,
       json_value.order_status,
       json_value.payment_method,
       json_value.total_amount,
       json_value.transaction_timestamp,
       json_value.customer_id,
       explode(array_distinct(json_value.items)) AS item
FROM gizmobox.silver.orders_json;
```

```python
%sql
SELECT
  order_id,
  order_status,
  payment_method,
  total_amount,
  transaction_timestamp,
  customer_id,
  item.item_id,
  item.name,
  item.price,
  item.quantity,
  item.category,
  item.details.brand,
  item.details.color
from
  gizmobox.silver.tv_item_exploded
```

```python
%sql
create table if not exists gizmobox.silver.orders as
SELECT
  order_id,
  order_status,
  payment_method,
  total_amount,
  transaction_timestamp,
  customer_id,
  item.item_id,
  item.name,
  item.price,
  item.quantity,
  item.category,
  item.details.brand,
  item.details.color
from
  gizmobox.silver.tv_item_exploded
```
